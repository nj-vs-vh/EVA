```python
from cr_knee_fit.elements import Element
```

```python
main_elements = [
    Element.H,
    Element.He,
    Element.Be,
    Element.C,
    Element.O,
    Element.Mg,
    Element.Si,
    Element.Fe,
    Element.Cu,
]
```

```python
from matplotlib.colors import LinearSegmentedColormap

ELEMENTS_CMAP = LinearSegmentedColormap.from_list(
    "elements",
    [(el.lnA_normalized, el.color) for el in main_elements],
)
```

```python
import numpy as np

np.log(1)
```

```python
from matplotlib import pyplot as plt

fig, ax = plt.subplots()


for elements, size in zip((main_elements, Element.regular()), (30, 10)):
    As = []
    Zs = []
    cols = []
    for element in elements:
        As.append(element.A)
        Zs.append(element.Z)
        cols.append(ELEMENTS_CMAP(element.lnA_normalized))

    ax.scatter(Zs, As, c=cols, s=size)

ax.set_xlabel("$ Z $")
ax.set_ylabel("$ \\langle A \\rangle $")
```
