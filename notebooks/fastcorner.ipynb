```python
import datetime
from pathlib import Path

import corner
import emcee

from cr_knee_fit.analysis import FitConfig
from cr_knee_fit.model import Model
from cr_knee_fit.utils import wrap_latex_labels
```

```python
outdir = Path("/home/nj-vs-vh/Documents/Science/cr-knee-fit/out/phase-6/2pop_kneebump")
backend = emcee.backends.HDFBackend(filename=str(outdir / "chain.h5"))
```

```python
tau = backend.get_autocorr_time(quiet=True)
print(tau)
print(tau.mean())
```

```python
sample = backend.get_chain(flat=True, thin=600, discard=1500)
assert sample is not None
print(f"Sample obtained: {sample.shape}")
```

```python
config_json = outdir / "config-dump.json"
config = FitConfig.model_validate_json(config_json.read_text())
```

```python
model_example = config.generate_initial_guess(None)
```

```python
ml_file = outdir / "mle-prelim.txt"
try:
    model_mle = Model.load(ml_file, layout_info=model_example.layout_info())
except Exception as e:  # noqa: BLE001
    print(e)
    model_mle = None
```

```python
fig = corner.corner(
    sample,
    labels=wrap_latex_labels(model_example.labels(latex=True)),
    show_titles=True,
)

if model_mle is not None:
    corner.overplot_lines(fig, model_mle.pack())
```

```python
fc_dir = outdir / "fastcorners"
fc_dir.mkdir(exist_ok=True)
fig.savefig(
    fc_dir
    / f"{datetime.datetime.now().isoformat(timespec='minutes')}_{backend.iteration}_steps.png"  # noqa: DTZ005
)
```
